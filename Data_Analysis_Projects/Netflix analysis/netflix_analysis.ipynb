```python
df['type'].value_counts()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
sns.countplot(x='type',data=df)
plt.title("Movies vs TV shows on Netflix")
plt.show()
```

```python
df['country'].value_counts().head(10)
```

```python
top_countries = df['country'].value_counts().head(10)
top_countries.plot(kind='bar',figsize=(10,5))
plt.xlabel("country")
plt.ylabel("Number of titles")
plt.show()
```

```python
df['release_year'].value_counts().sort_index()
```

```python
year_counts= df['release_year'].value_counts().sort_index()
year_counts.plot(kind='line',figsize=(10,5))
plt.title("Netflix Contents Growth Over time")
plt.xlabel("Year")
plt.ylabel("Number of Titles")
plt.show()
```

```python
type_year = df.groupby(['release_year','type']).size().unstack()
type_year.plot(figsize=(10,5))
plt.title("Movies vs TV shows Released each Year")
plt.xlabel("Year")
plt.ylabel("Number of Titles")
plt.show()
```

```python
df['listed_in'].value_counts().head(10)
```

```python
top_genres = df['listed_in'].value_counts().head(10)
top_genres.plot(kind='bar' ,figsize=(10,5) , color="#003049")
plt.title("Top 10 Genres on Netflix",fontsize=15 , color='#853953')
plt.xlabel("Genre")
plt.ylabel("Number of titles")
plt.show()
```

```python
df['rating'].value_counts().head(10)
```

```python
df['rating'].value_counts().head(10).plot(kind='bar', figsize=(10,5), color='#5E0006')
plt.title("Top Content Ratings on Netflix", fontsize= 12 , color='#4B2E2B')
plt.xlabel("Rating",color="#4B2E2B")
plt.ylabel("Number of titles" ,color='#4B2E2B')
plt.show()
```
